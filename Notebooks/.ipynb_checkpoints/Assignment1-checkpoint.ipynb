```python
# 3.1 Normalise
```

```python
def normalise(values, minimum=None, maximum=None):
    
    if not values:
        return []

    if minimum is None:
        minimum = min(values)

    if maximum is None:
        maximum = max(values)

    # Avoid division by zero when all values are identical.
    
    if maximum == minimum:
        return [0.0 for value in values]

    return [
        (value - minimum) / (maximum - minimum)
        for value in values
    ]
```

```python
values = [12, 18, 25, 40]
print(normalise(values))

same_values = [5, 5, 5, 5]
print(normalise(same_values))
```

```python
# Summarise_scores
```

```python
def summarise_scores(scores):
 
    if not scores:
        return {
            "count": 0,
            "mean": None,
            "minimum": None,
            "maximum": None,
            "above_threshold": 0
        }

    return {
        "count": len(scores),
        "mean": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
        "above_threshold": sum(score >= 0.8 for score in scores)
    }
```

```python
scores_1 = [0.65, 0.78, 0.81, 0.92, 0.88]
scores_2 = []

print(summarise_scores(scores_1))
print(summarise_scores(scores_2))
```

```python
# safe-divide
```

```python
def safe_divide(numerator, denominator, default=0.0):
    try:
        return numerator / denominator
    except (ZeroDivisionError, TypeError):
        return default
```

```python
print("Normal:", safe_divide(10, 2))
print("Zero denominator:", safe_divide(10, 0))
print("Text input:", safe_divide("ten", 2))
```

```python
import numpy as np
```

```python
# Oone- dimensional array

numbers = np.array([12, 18, 25, 31, 40, 47])

print("Array:", numbers)
print("Mean:", numbers.mean())
print("Standard deviation:", numbers.std())
print("Minimum:", numbers.min())
print("Maximum:", numbers.max())

# Two dimensional array witn (3,4) Shape

matrix = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [20, 30, 40, 50]
])

print("Matrix:")
print(matrix)

print("Shape:", matrix.shape)

transposed = matrix.T

print("Transpose shape:", transposed.shape)

# Broadcasting

column_means = matrix.mean(axis=0)

print("Column means:")
print(column_means)

centered_matrix = matrix - column_means

print("Matrix after subtracting column means:")
print(centered_matrix)
```

```python
data = {
    "model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost",
        "SVM",
        "Neural Network"
    ],
    "accuracy": [0.78, 0.81, 0.89, 0.93, np.nan, 0.91],
    "training_time": [2.1, 1.8, 5.2, 6.4, 4.5, 12.3],
    "epochs": [10, 10, 20, 25, 15, 30]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Filter and Sort
high_accuracy_models = df[df["accuracy"] >= 0.85]

high_accuracy_models = high_accuracy_models.sort_values(
    by="accuracy",
    ascending=False
)

print(high_accuracy_models)
```
